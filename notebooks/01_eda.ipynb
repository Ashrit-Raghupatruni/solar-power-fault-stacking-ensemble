```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.config import STATION_FILES, RAW_SENSOR_FEATURES, CATEGORY_LABELS, SEVERITY_LABELS

print('Configurations and dependencies successfully loaded!')
```

## 1. Load Data Sample from Station 1 (Laboratory Baseline)

```python
df_sample = pd.read_csv(STATION_FILES[1], nrows=50000)
print('Sample shape:', df_sample.shape)
df_sample[['timestamp', 'voltage_V', 'current_A', 'power_W', 'frequency_Hz', 'power_factor', 'fault_category', 'severity']].head(10)
```

## 2. Fault Category & Severity Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
df_sample['fault_category'].value_counts().plot(kind='bar', ax=axes[0], color='royalblue', title='Fault Categories')
axes[0].set_ylabel('Sample Count')
axes[0].grid(axis='y', alpha=0.3)

df_sample['severity'].value_counts().plot(kind='bar', ax=axes[1], color='crimson', title='Fault Severity Grades')
axes[1].set_ylabel('Sample Count')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 3. Physical Sensor Telemetry Correlation Matrix

```python
plt.figure(figsize=(9, 7))
corr = df_sample[RAW_SENSOR_FEATURES].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Electrical Telemetry Correlation Heatmap')
plt.show()
```
